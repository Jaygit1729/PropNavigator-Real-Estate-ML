**Input:** `data/fe/featured_properties.csv`  (39,621 rows × 48 cols)

**Goal:** understand distributions, outliers, and feature–price relationships before
making preprocessing decisions.

## 1. Target Variable — `price_in_cr`

Understanding the target distribution is the first step. A heavily skewed target
usually benefits from a log transform at modelling time.

```python
fig = make_subplots(
    rows=1, cols=3,
    subplot_titles=('price_in_cr (raw)',
                    'price_in_cr (clipped at 20 Cr)',
                    'log(1 + price_in_cr)'))

fig.add_trace(go.Histogram(x=df['price_in_cr'], nbinsx=100,
                           marker_color=PRIMARY), row=1, col=1)
fig.add_trace(go.Histogram(x=df['price_in_cr'].clip(upper=20), nbinsx=100,
                           marker_color=PRIMARY), row=1, col=2)
fig.add_trace(go.Histogram(x=np.log1p(df['price_in_cr']), nbinsx=100,
                           marker_color=ACCENT), row=1, col=3)

fig.update_layout(title_text='Price Distribution', showlegend=False, height=380)
fig.show()

print(df['price_in_cr'].describe())
print(f"raw skew : {df['price_in_cr'].skew():.2f}")
print(f"log skew : {np.log1p(df['price_in_cr']).skew():.2f}")
```

> **Conclusion.** `price_in_cr` is extremely right-skewed — median **2.8 Cr** but a
> max of **1000 Cr** drives skew to **71**. Clipping at 20 Cr shows the real mass sits
> between ~1–5 Cr. The `log(1+price)` view is near-symmetric (**skew ≈ 0.96**).
> → **Action:** model on **log price**, and treat the extreme upper tail as outliers
> (see §11).

```python
df[df['price_in_cr'] >30]
```

```python
fig = px.box(df, x='property_type', y='price_in_cr', color='property_type',
             points=False, title='Price by Property Type (outliers hidden)')
fig.update_yaxes(range=[0, 20])
fig.update_layout(showlegend=False, height=420)
fig.show()

df.groupby('property_type')['price_in_cr'].describe()
```

> **Conclusion.** Price scales cleanly with build type: **Flat** median **2.5 Cr**
> (n=28,433) < **Independent Builder Floor** **3.3 Cr** (n=9,168) <

> **Independent House** **6.6 Cr** (n=2,020). Flats dominate supply (~72%).

> → **Action:** `property_type` is a strong, non-redundant predictor — **keep it**.

## 7. Amenity & Overlooking Flags vs Price

For each binary flag, compare median price when the flag is 1 vs 0. A large gap
suggests price signal — **but watch for confounding by property type.**

```python
def lift_table(cols, on, off):
    rows = []
    for c in cols:
        m1 = df[df[c] == 1]['price_in_cr'].median()
        m0 = df[df[c] == 0]['price_in_cr'].median()
        rows.append({'feature': c, on: m1, off: m0,
                     'lift_pct': round((m1 - m0) / m0 * 100, 1)})
    return pd.DataFrame(rows).sort_values('lift_pct', ascending=False)

amenity_df = lift_table(amenity_cols, 'has_feature', 'no_feature')
amenity_df
```

```python
fig = go.Figure()
fig.add_trace(go.Bar(name='has feature', x=amenity_df['feature'],
                     y=amenity_df['has_feature'], marker_color=PRIMARY))
fig.add_trace(go.Bar(name='no feature', x=amenity_df['feature'],
                     y=amenity_df['no_feature'], marker_color='#A9CCE3'))
fig.update_layout(barmode='group', height=430, xaxis_tickangle=30,
                  yaxis_title='median price_in_cr',
                  title='Median Price: Has Amenity vs Does Not')
fig.show()
```

> **Conclusion (important nuance).** Only `has_servant_room` (**+50%**) and `has_ac`
> (**+43%**) show a clean positive lift. Counter-intuitively `has_lift` (**−32%**),
> `has_gym` (**−17%**) and `has_pool` (**−4%**) show **negative** raw lift — **not**
> because they reduce value, but because they cluster in **Flats**, the cheapest build
> type. This is **Simpson's paradox / confounding**. → **Action:** keep `has_servant_room`
> and `has_ac` as clear signals; treat the tower-amenity flags as **property-type proxies**
> and judge them *within* type (or via the model), not by raw lift.

```python
ov_df = lift_table(ov_cols, 'overlooks', 'does_not')
ov_df
```

> **Conclusion.** Every overlooking flag shows near-zero or **negative** raw lift
> (`ov_sea` the most negative at **−17%**, again a Flat-cluster artefact). None is a
> strong standalone driver. → **Action:** low priority; keep as-is but don't expect
> importance. *(This corrects the older summary that called `ov_sea` the top-lift flag.)*

## 8. Location — distance-to-landmark features

The GPS coordinates are cleaned and turned into **distance-to-landmark** features back in
**feature engineering** (`src/feature_engineering/feature_eng.py`) — the coordinate repair,
the swap-fix, and the distance maths all live there. So `featured_properties.csv` already
contains them. Here we simply *check whether they carry price signal* — we don't rebuild
anything.

The four landmarks point in different directions — the **office hub** (Cyber City), the
**luxury belt** (Golf Course Road), the **airport**, and the **cheap industrial edge**
(Manesar) — so together they describe where any home sits.

```python
# These columns arrive ready-made from feature engineering.
dist_cols = ['dist_to_cyber_city', 'dist_to_golf_road', 'dist_to_airport', 'dist_to_manesar']

geo = df[df['dist_to_cyber_city'].notna()]   # ~187 rows had un-fixable coords -> skip them
print("rows with a valid location:", len(geo))
geo[dist_cols].describe().round(1)
```

### Does being closer actually mean pricier?

```python
# Target is PRICE. We relate each distance to log(price) — 'log' just squeezes the
# few sky-high prices so a handful of mansions don't distort the number.
log_price = np.log(geo['price_in_cr'])

# Reading the sign:  negative = closer is pricier,  positive = farther is pricier.
for col in dist_cols:
    print(f"{col:20s}: {geo[col].corr(log_price):+.2f}")
```

> **What the numbers say** (correlation with `log(price)`):
> - Closer to **Golf Course Road** (−0.48), **Cyber City** (−0.47), or the **Airport**
>   (−0.45) → **more expensive**.
> - **Farther** from **Manesar** (+0.34) → **more expensive** (it's the cheap edge, so
>   "far from Manesar" means "deep in premium Gurgaon").
>
> Distance is **one of the strongest predictors** in the data — on par with bedroom count —
> and the strongest *location* signal.

### Price falls as you move away from the hub

```python
geo = geo.assign(distance_band=pd.cut(
    geo['dist_to_cyber_city'], [0, 5, 10, 15, 100],
    labels=['0-5 km', '5-10 km', '10-15 km', '15+ km']))

ring = geo.groupby('distance_band', observed=True)['price_in_cr'].median().round(2)

fig = px.bar(x=ring.index.astype(str), y=ring.values, text=ring.values,
             title='Typical price (Cr) by distance from the office hub')
fig.update_traces(marker_color=PRIMARY, textposition='outside')
fig.update_layout(height=400, xaxis_title='distance from Cyber City',
                  yaxis_title='median price (Cr)')
fig.show()
ring
```

> **The headline:** a clean staircase — **5.0 Cr within 5 km of the hub, down to 1.85 Cr
> beyond 15 km.** The farther out, the cheaper. The strongest location story in the data.

### The same thing on a map

```python
sample = geo.sample(5000, random_state=42)
fig = px.scatter(sample, x='longitude', y='latitude', color='price_in_cr',
                 color_continuous_scale='Turbo', range_color=[1, 8],
                 title='Where the expensive homes are (redder = pricier)')
fig.update_traces(marker_size=4, opacity=0.5)
fig.update_layout(height=520)
fig.show()
```

> **What the map shows:** the red (expensive) dots cluster toward the centre-east —
> Cyber City and the Golf Course Road belt — and fade to blue toward the outer edges. The
> map agrees with the numbers.

> **Why only four landmarks?** Landmarks that sit close together give almost the same
> distance for every home — the same information twice. A few landmarks in *different
> directions* are enough to locate any home; more just repeat what we already know.

## 9. Correlation Heatmap

```python
num_cols = [
    'price_in_cr', 'price_per_sqft', 'area', 'bedRoom', 'bathroom', 'balcony',
    'floornum', 'total_floor', 'covered_parking', 'open_parking', 'total_parking',
    'is_corner',
    'has_ac', 'has_power_backup', 'has_gym', 'has_pool', 'has_club',
    'has_lift', 'has_servant_room', 'is_gated',
    'ov_park', 'ov_pool', 'ov_club', 'ov_main_road', 'ov_sea', 'ov_others',
    # location features engineered from GPS (see the Location section above)
    'dist_to_cyber_city', 'dist_to_golf_road', 'dist_to_airport', 'dist_to_manesar',
]
corr = df[num_cols].corr()
mask = np.triu(np.ones_like(corr, dtype=bool))
corr_masked = corr.mask(mask)

fig = px.imshow(corr_masked, text_auto='.2f', color_continuous_scale='RdBu_r',
                zmin=-1, zmax=1, aspect='auto',
                title='Correlation Matrix — Numerical Features')
fig.update_layout(height=760, width=900)
fig.update_traces(textfont_size=8)
fig.show()
```

## 11. Outlier Investigation

These boxplots directly inform the IQR bounds used in preprocessing.

```python
fig = make_subplots(rows=1, cols=3,
    subplot_titles=('price_in_cr', 'area', 'price_per_sqft'))
for i, col in enumerate(['price_in_cr', 'area', 'price_per_sqft'], start=1):
    fig.add_trace(go.Box(y=df[col].dropna(), name=col, boxpoints='outliers',
                         marker=dict(size=2, opacity=0.3),
                         marker_color=PRIMARY, showlegend=False), 1, i)
fig.update_layout(height=460, title_text='Outlier Boxplots (IQR method)')
fig.show()
```

```python
print('=== IQR Bounds Summary (copy into preprocessing) ===')
for col in ['price_in_cr', 'area', 'price_per_sqft']:
    Q1, Q3 = df[col].quantile(0.25), df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower, upper = Q1 - 1.5 * IQR, Q3 + 1.5 * IQR
    n_out = ((df[col] < lower) | (df[col] > upper)).sum()
    print(f'{col:20s}  lower={lower:9.2f}  upper={upper:9.2f}  outliers={n_out}')
```

> **Conclusion.** All three targets carry a heavy upper tail. The printed IQR bounds are
> the concrete thresholds to apply. → **Action:** in preprocessing, remove rows outside
> the IQR fences for `price_in_cr`, `area`, `price_per_sqft` (after the `price_per_sqft`
> floor of 1,000), *before* the log transform.
