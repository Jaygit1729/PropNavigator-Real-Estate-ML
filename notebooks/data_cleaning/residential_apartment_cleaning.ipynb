## Removing Duplicate Listings

### What is the problem?
On 99acres, the same flat is often posted by several agents. So one real flat shows up as
many rows in our data. We want **one row per real flat**.

If we leave the duplicates in:
- **The score becomes fake.** The same flat can land in both training and test data, so the
  model has already "seen" the answer and looks better than it really is.
- **Popular flats get too much weight.** A flat posted 10 times pulls the model towards itself.

### Why is it tricky?
There is no flat number in our data (that is only on the detail page, which we did not scrape).
So we cannot be 100% sure two rows are the exact same flat. We have to decide using the columns
we do have. This means any rule we make can go wrong in two ways:
- Too strict -> real duplicates stay in.
- Too loose -> two different flats get merged into one (we lose real data).

So the goal is simply to make as few wrong calls as possible.

### How did we decide?
We tried different combinations of columns to define "same flat", then actually **looked at the
rows** each rule grouped together — and even opened a few listings on 99acres to check.

What we found:
- When we **left price out** of the rule, it merged flats whose carpet area and price were
  clearly different (for example, area 1100 vs 1400 sqft, price 2.05 vs 2.24 Cr). Those are
  **different flats**, so merging them throws away real information.
- When we **kept price in** the rule, the grouped rows had the same price, area and floor, and
  only small noise differed (a tiny carpet-area difference, or one agent writing a different
  facing). Those really are **the same flat reposted**.

The simple takeaway: **if two rows have the same details and the same price, treat them as the
same flat.** Price is the most reliable signal we have.

### What we use
We mark two rows as the same flat when these all match:
`property_name, society, price_in_cr, areaWithType, floorNum, facing, overlooking`

- **Price** is the strongest clue.
- **Society, area, floor, facing, overlooking** pin down the specific unit.
- We **leave out** columns agents fill inconsistently (bathroom, carpet area, furnishing) — using
  them would split the same flat into "different" rows and miss real duplicates.

### This is easy to change later
This rule is a safe, careful choice. After we build the model:
- If the score looks **too good to be true**, duplicates may still be leaking -> make the rule
  stricter.
- If it looks reasonable, we keep it.

It is a one-line change (`DEDUP_KEY` below), so we can adjust it any time.

```python
from IPython.display import display

# Work on a copy of the cleaned data
df_dedupe = column_cleaning.copy()

# --- Chosen dedup key (Option A, refined) ---
# Rows with identical values for ALL these columns are treated as the SAME flat,
# reposted by different agents.
DEDUP_KEY = ['property_name', 'society', 'price_in_cr',
             'areaWithType', 'floorNum', 'facing', 'overlooking']

# How many rows would this key remove?
n_drop = df_dedupe.duplicated(subset=DEDUP_KEY).sum()
print(f"Total rows        : {len(df_dedupe)}")
print(f"Duplicates to drop: {n_drop} ({100 * n_drop / len(df_dedupe):.1f}%)")
```

### Sanity check — look at the duplicate groups

Each group below is a set of rows that share the same `DEDUP_KEY`. Eyeball them: if the rows
are clearly the same flat (only price/agent text differs) the key is doing its job.

```python
# Columns that make it easy to judge if two rows are the same flat
VIEW = ['society', 'areaWithType', 'floorNum', 'facing', 'overlooking',
        'bedRoom', 'bathroom', 'carpetArea', 'price_in_cr', 'property_name', 'link']

def show_duplicates(key=DEDUP_KEY, n_groups=8):
    """Print the first few groups of rows flagged as duplicates by `key`."""
    dupes = df_dedupe[df_dedupe.duplicated(subset=key, keep=False)]
    n_total = dupes.groupby(key, dropna=False).ngroups
    print(n_total, "duplicate groups,", len(dupes), "rows involved")
    print()
    for i, (_, group) in enumerate(dupes.groupby(key, dropna=False), start=1):
        if i > n_groups:
            break
        print(f"--- group {i}  ({len(group)} rows) ---")
        display(group[VIEW])

show_duplicates()
```

### Apply deduplication

```python
before = len(column_cleaning)
flats_dedup = column_cleaning.drop_duplicates(subset=DEDUP_KEY, keep='first').reset_index(drop=True)
after = len(flats_dedup)

print(f"{before} -> {after} rows  ({before - after} duplicate repostings dropped)")
flats_dedup.head()
```
