```python
column_cleaning = (
    df_house
    .assign(
        society=lambda df_: df_['society']
            .str.strip()
            .str.lower()
    )

    .assign(
        price_in_cr=lambda df_: df_['price_in_cr']
            .str.strip()
    )

    .loc[lambda df_: (
        df_['price_in_cr'].notna() &
        (df_['price_in_cr'] != 'Price on Request') &
        (~df_['price_in_cr'].str.contains('-', na=False))   # drop range listings (multi-unit projects)
    )]

    .assign(
        price_in_cr=lambda df_: round(
            pd.to_numeric(
                df_['price_in_cr'].apply(
                    lambda value:
                        float(str(value).replace('Cr', '').strip())
                        if 'Cr' in str(value)
                        else float(str(value).replace('Lac', '').strip()) / 100
                )
            ),
            4
        )
            )
    .assign(
        price_per_sqft=lambda df_: (
            df_['price_per_sqft']
            .str.split('₹')
            .str.get(1)
            .str.split('/')
            .str.get(0)
            .str.replace('L', '', regex=False)
            .str.replace(',', '', regex=False)
            .str.strip()
            .pipe(pd.to_numeric, errors='coerce')
            .mul(
                np.where(
                    df_['price_per_sqft'].str.contains('L', na=False),
                    100000,
                    1
                )
            )
            .div(
                np.where(
                    df_['price_per_sqft'].str.contains('/sqyd', case=False, na=False),
                    9,
                    1
                )
            )
            .round()
            .astype('Int64')
        )
    )

)
```

### Final Cleaning Pipeline


```python
AGE_POSSESSION_MAP = {
    0: 'Undefined', 1: '1 to 5 Year Old', 2: '5 to 10 Year Old',
    3: '10+ Year Old', 5: 'Under Construction', 6: '0 to 1 Year Old'
}


column_cleaning = (
    df_house
    .rename(columns={
    'price': 'price_in_cr',
    'area': 'price_per_sqft'})

    .assign(
        society=lambda df_: df_['society']
            .str.strip()
            .str.lower()
            )

    .assign(
        price_in_cr=lambda df_: df_['price_in_cr']
            .str.strip()
            )

    .loc[lambda df_: (
        df_['price_in_cr'].notna() &
        (df_['price_in_cr'] != 'Price on Request') &
        (~df_['price_in_cr'].str.contains('-', na=False))   # drop range listings (multi-unit projects)
                    )]

    .assign(
        price_in_cr=lambda df_: round(
            pd.to_numeric(
                df_['price_in_cr'].apply(
                    lambda value:
                        float(str(value).replace('Cr', '').strip())
                        if 'Cr' in str(value)
                        else float(str(value).replace('Lac', '').strip()) / 100
                )
            ),
            4
        )
            )
    .assign(
        price_per_sqft=lambda df_: (
            df_['price_per_sqft']
            .str.split('₹')
            .str.get(1)
            .str.split('/')
            .str.get(0)
            .str.replace('L', '', regex=False)
            .str.replace(',', '', regex=False)
            .str.strip()
            .pipe(pd.to_numeric, errors='coerce')
            .mul(
                np.where(
                    df_['price_per_sqft'].str.contains('L', na=False),
                    100000,
                    1
                )
            )
            .div(
                np.where(
                    df_['price_per_sqft'].str.contains('/sqyd', case=False, na=False),
                    9,
                    1
                )
            )
            .round()
            .astype('Int64')
        )
    
    , facing=lambda df_: df_['facing']
                                .replace('0', "Not Available")
                                .str
                                .lower()
    , furnishing=lambda df_: df_['furnishing']
                                .replace('0', "Not Available")
                                .str
                                .lower()
    , agePossession=lambda df_: df_['agePossession'].map(AGE_POSSESSION_MAP).fillna('Undefined')
    ))

print(f'Shape: {column_cleaning.shape}')
print(f'\nNull counts:')
print(column_cleaning.isnull().sum())
column_cleaning.head()

```
