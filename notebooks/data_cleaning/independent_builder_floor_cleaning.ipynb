```python
column_cleaning = (
    df_bf
    .assign(
        society=lambda df_: df_['society']
            .str.strip()
            .str.lower()
    )
    .assign(
        price_in_cr=lambda df_: df_['price_in_cr']
            .str.strip()
    )
    .loc[
        lambda df_: (
            df_['price_in_cr'].notna()
            & (df_['price_in_cr'] != 'Price on Request')
            & (df_['price_in_cr'] != '70,000')
        )
    ]
    .assign(
        price_in_cr=lambda df_: round(
            pd.to_numeric(
                df_['price_in_cr'].apply(
                    lambda value:
                    float(str(value).replace('Cr', '').strip())
                    if 'Cr' in str(value)
                    else float(str(value).replace('Lac', '').strip()) / 100
                )
            ),
            2
        )
    )
    .assign(
        price_per_sqft=lambda df_: (
            df_['price_per_sqft']
            .str.split('₹')
            .str.get(1)
            .str.split('/')
            .str.get(0)
            .str.replace('L', '', regex=False)
            .str.replace(',', '', regex=False)
            .str.strip()
            .pipe(pd.to_numeric, errors='coerce')
            .mul(
                np.where(
                    df_['price_per_sqft'].str.contains('L', na=False),
                    100000,
                    1
                )
            )
            .div(
                np.where(
                    df_['price_per_sqft'].str.contains('/sqyd', case=False, na=False),
                    9,
                    1
                )
            )
            .round()
            .astype('Int64')
        )
    )
)
```

### Final Cleaning Pipeline


```python
AGE_POSSESSION_MAP = {
    1: '1 to 5 Year Old', 2: '5 to 10 Year Old',
    3: '10+ Year Old', 5: 'Under Construction', 6: '0 to 1 Year Old'
}

column_cleaning = (
    df_bf
    .rename(columns={
    'price': 'price_in_cr',
    'area': 'price_per_sqft'
})

    .assign(
        society=lambda df_: df_['society']
            .str.strip()
            .str.lower()
    )
    .assign(
        price_in_cr=lambda df_: df_['price_in_cr']
            .str.strip()
    )
    .loc[
        lambda df_: (
            df_['price_in_cr'].notna()
            & (df_['price_in_cr'] != 'Price on Request')
            & (df_['price_in_cr'] != '70,000')
        )
    ]
    .assign(
        price_in_cr=lambda df_: round(
            pd.to_numeric(
                df_['price_in_cr'].apply(
                    lambda value:
                    float(str(value).replace('Cr', '').strip())
                    if 'Cr' in str(value)
                    else float(str(value).replace('Lac', '').strip()) / 100
                )
            ),
            2
        )
    )
    .assign(
        price_per_sqft=lambda df_: (
            df_['price_per_sqft']
            .str.split('₹')
            .str.get(1)
            .str.split('/')
            .str.get(0)
            .str.replace('L', '', regex=False)
            .str.replace(',', '', regex=False)
            .str.strip()
            .pipe(pd.to_numeric, errors='coerce')
            .mul(
                np.where(
                    df_['price_per_sqft'].str.contains('L', na=False),
                    100000,
                    1
                )
            )
            .div(
                np.where(
                    df_['price_per_sqft'].str.contains('/sqyd', case=False, na=False),
                    9,
                    1
                )
            )
            .round()
            .astype('Int64')
        )
    
    , facing=lambda df_: df_['facing']
                                .replace('0', "Not Available")
                                .str
                                .lower()
    , furnishing=lambda df_: df_['furnishing']
                                .replace('0', "Not Available")
                                .str
                                .lower()
    , agePossession=lambda df_: df_['agePossession'].map(AGE_POSSESSION_MAP).fillna('Undefined')
    ))

print(f'Shape: {column_cleaning.shape}')
print(f'\nNull counts:')
print(column_cleaning.isnull().sum())
column_cleaning.head()
```

```python
from IPython.display import display

# Work on a copy of the cleaned data
df_dedupe = column_cleaning.copy()

# --- Chosen dedup key (Option A, refined) ---
# Rows with identical values for ALL these columns are treated as the SAME flat,
# reposted by different agents.
DEDUP_KEY = ['property_name', 'society', 'price_in_cr',
             'areaWithType', 'floorNum', 'facing', 'overlooking']

# How many rows would this key remove?
n_drop = df_dedupe.duplicated(subset=DEDUP_KEY).sum()
print(f"Total rows        : {len(df_dedupe)}")
print(f"Duplicates to drop: {n_drop} ({100 * n_drop / len(df_dedupe):.1f}%)")
```

```python
# Columns that make it easy to judge if two rows are the same flat
VIEW = ['society', 'areaWithType', 'floorNum', 'facing', 'overlooking',
        'bedRoom', 'bathroom', 'carpetArea', 'price_in_cr', 'property_name', 'link']

def show_duplicates(key=DEDUP_KEY, n_groups=8):
    """Print the first few groups of rows flagged as duplicates by `key`."""
    dupes = df_dedupe[df_dedupe.duplicated(subset=key, keep=False)]
    n_total = dupes.groupby(key, dropna=False).ngroups
    print(n_total, "duplicate groups,", len(dupes), "rows involved")
    print()
    for i, (_, group) in enumerate(dupes.groupby(key, dropna=False), start=1):
        if i > n_groups:
            break
        print(f"--- group {i}  ({len(group)} rows) ---")
        display(group[VIEW])

show_duplicates()
```
